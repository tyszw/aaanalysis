# Variable JMD length / partial part-based input verification

This notebook verifies recently added behaviors for `SequenceFeature.get_df_parts`:

1. per-sample `jmd_n_len` / `jmd_c_len` columns
2. partial part-based formats (`entry+tmd`, `entry+jmd_n+tmd`, `entry+tmd+jmd_c`)


```python
import aaanalysis as aa
import pandas as pd
```

```python
sf = aa.SequenceFeature()
df_seq = aa.load_dataset(name='DOM_GSEC', n=5)
df_seq.head(2)
```

```python
# 1) per-sample row-wise JMD lengths
df_var = df_seq[['entry', 'sequence']].copy()
n = len(df_var)
df_var['jmd_n_len'] = [2 + (i % 5) for i in range(n)]
df_var['jmd_c_len'] = [1 + (i % 5) for i in range(n)]

df_parts_var = sf.get_df_parts(
    df_seq=df_var,
    list_parts=['jmd_n', 'tmd', 'jmd_c'],
    jmd_n_len=None,
    jmd_c_len=None,
)

assert (df_parts_var['jmd_n'].str.len().to_numpy() == df_var['jmd_n_len'].to_numpy()).all()
assert (df_parts_var['jmd_c'].str.len().to_numpy() == df_var['jmd_c_len'].to_numpy()).all()
df_parts_var.head(2)
```

```python
# 2-a) partial part-based: entry + tmd
df_tmd_only = df_seq[['entry', 'tmd']].copy()
df_parts_tmd_only = sf.get_df_parts(
    df_seq=df_tmd_only,
    list_parts=['jmd_n', 'tmd', 'jmd_c'],
    jmd_n_len=None,
    jmd_c_len=None,
)
assert (df_parts_tmd_only['jmd_n'].str.len() == 0).all()
assert (df_parts_tmd_only['jmd_c'].str.len() == 0).all()
df_parts_tmd_only.head(2)
```

```python
# 2-b) partial part-based: entry + jmd_n + tmd
df_jmdn_tmd = df_seq[['entry', 'jmd_n', 'tmd']].copy()
df_parts_jmdn_tmd = sf.get_df_parts(
    df_seq=df_jmdn_tmd,
    list_parts=['jmd_n', 'tmd', 'jmd_c'],
    jmd_n_len=None,
    jmd_c_len=None,
)
assert (df_parts_jmdn_tmd['jmd_c'].str.len() == 0).all()
df_parts_jmdn_tmd.head(2)
```

```python
# 2-c) partial part-based: entry + tmd + jmd_c
df_tmd_jmdc = df_seq[['entry', 'tmd', 'jmd_c']].copy()
df_parts_tmd_jmdc = sf.get_df_parts(
    df_seq=df_tmd_jmdc,
    list_parts=['jmd_n', 'tmd', 'jmd_c'],
    jmd_n_len=None,
    jmd_c_len=None,
)
assert (df_parts_tmd_jmdc['jmd_n'].str.len() == 0).all()
df_parts_tmd_jmdc.head(2)
```
